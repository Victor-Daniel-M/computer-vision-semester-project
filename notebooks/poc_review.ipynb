# Enhancement + Recognition POC

Quick review notebook for the frame extraction, enhancement outputs, grids, and weak-label recognition baseline.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
OUTPUT_DIR = ROOT / 'outputs' / 'poc'
frame_index = pd.read_csv(OUTPUT_DIR / 'frame_index.csv')
summary = json.loads((OUTPUT_DIR / 'reports' / 'summary.json').read_text())
frame_index.head()
```

```python
summary['videos'], summary['frames'], summary['pipelines']
```

```python
for grid in summary['grids'][:4]:
    display(Image(filename=str(ROOT / grid)))
```

```python
rows = []
for variant, models in summary['metrics'].items():
    for model_name, result in models.items():
        rows.append({'variant': variant, 'model': model_name, 'accuracy': result['accuracy']})
pd.DataFrame(rows).sort_values(['model', 'accuracy'], ascending=[True, False])
```
